```python
from google.colab import files
import os

# آپلود فایل
def upload_file():
    print("Please upload an image file:")
    uploaded = files.upload()  # آپلود فایل
    if not uploaded:
        print("No file uploaded. Exiting...")
        return None
    image_path = list(uploaded.keys())[0]  # گرفتن نام فایل آپلود شده
    print(f"Uploaded file: {image_path}")
    return image_path

# آپلود تصویر
image_path = upload_file()

```
